# Dataset 2 – Basic File Validation

```python
import pandas as pd, pathlib
root = pathlib.Path('.').resolve()
csv_path = root / 'data/raw/dataset_2/Persons Engaged in Industry by Enterprise Type – PLFS Year.csv'
df = pd.read_csv(csv_path)
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
print('Column names:', list(df.columns))
print('Data types:', df.dtypes)
print('Blank rows:', df.isnull().all(axis=1).sum())
print('Duplicated rows:', df.duplicated(keep=False).sum())
display(df.head())
display(df.tail())
```
