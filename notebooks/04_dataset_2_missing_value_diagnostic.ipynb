# Dataset 2 – Missing‑Value Diagnostic

```python
import pandas as pd, pathlib, numpy as np
root = pathlib.Path('.').resolve()
csv_path = root / 'data/raw/dataset_2/Persons Engaged in Industry by Enterprise Type – PLFS Year.csv'
df = pd.read_csv(csv_path)
n_rows, n_cols = df.shape
print('Rows:', n_rows, 'Columns:', n_cols)
num_fields = [
    'Persons Engaged In The Industry Groups By Enterprise Type (%) (UOM:%(Percentage)), Scaling Factor:1',
    'Estimated Persons (UOM:Number), Scaling Factor:100',
    'Sample Number Of Workers In The Enterprise (UOM:Number), Scaling Factor:1'
]
# Field‑level missing summary
field_summary = []
for col in df.columns:
    miss = df[col].isnull().sum()
    field_summary.append({
        'field': col,
        'missing_count': miss,
        'missing_pct': round(miss/n_rows*100,2),
        'non_missing_count': n_rows-miss,
        'non_missing_pct': round(100-miss/n_rows*100,2)
    })
field_df = pd.DataFrame(field_summary)
display(field_df)
# Missingness patterns across numeric fields
patterns = df[num_fields].isnull().astype(int).apply(lambda r: ''.join(r.astype(str)), axis=1)
pattern_counts = patterns.value_counts().reset_index()
pattern_counts.columns = ['pattern','count']
pattern_counts['percentage'] = (pattern_counts['count']/n_rows*100).round(2)
display(pattern_counts)
# Distribution of records with any numeric missing
any_missing = df[num_fields].isnull().any(axis=1)
missing_df = df[any_missing]
for dim in ['Year','State','Gender','Type Of Areas','Industry Division Type','Enterprise Type']:
    print(f'
Missing by {dim}:')
    print(missing_df[dim].value_counts())
```
